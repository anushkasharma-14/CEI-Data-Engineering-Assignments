### Setup and Data Loading  
Import Spark libraries, create a Spark Session, and load the CSV and Parquet dataset for basic exploration.

```python
# Import required PySpark libraries
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
# Create Spark Session
spark = SparkSession.builder \
    .appName("Week6 Spark Assignment") \
    .getOrCreate()
```

### Q3 - Load CSV File  
Read "data/source.csv" with header and inferSchema enabled.

```python
# Load CSV file into DataFrame
df = spark.read.csv("data/source.csv", header=True, inferSchema=True)
```

```python
# Load Parquet file into DataFrame
parquet_df = spark.read.parquet("data/source.parquet")
```

```python
# Display first 5 rows of CSV data
df.show(5)
```

```python
# Display first 5 rows of Parquet data
parquet_df.show(5)
```

```python
# Print schema of CSV DataFrame
df.printSchema()
```

```python
# Count total number of rows in CSV DataFrame
df.count()
```

```python
# Show column names in DataFrame
df.columns
```

```python
# Get distinct values from 'category' column
df.select("category").distinct().show()
```

### Q5 - Filter and Select Columns  
Select product_id and price where category is 'Electronics'.

```python
# Select only the required columns
electronics_df = df.filter(col("category") == "Electronics").select("product_id", "price")

# Display the result
electronics_df.show(5)
```

### Q6 - DataFrame Update  
Rename column old_name("supplier") to new_name("vendor") and cast price column to Double.

```python
# Spark automatically converted 'price' to Double because we used inferSchema=True.
# For this question, we first convert it to String and then cast it back to Double
# to demonstrate the required operation.

df_q6 = df.withColumn("price", col("price").cast("string"))

# Rename the "supplier" column to "vendor"
# Convert price back to Double

df_q6 = df_q6.withColumnRenamed("supplier", "vendor") \
    .withColumn("price", col("price").cast("double"))

# Display the updated schema
df_q6.printSchema()

# Display a few records
df_q6.show(5)
```

### Q8 - Conditional Filtering  
Filter df_orders where status is 'Completed' and amount > 1000.

```python
# Filter rows where status is 'Completed' and amount is greater than 1000
completed_orders = df.filter((col("status") == "Completed") & (col("amount") > 1000))

# Display the filtered records
completed_orders.show(5)
```

### Q10 - Derived Column  
Create final_price as base_price * 1.18 (18% tax).

```python
# Create a new column by adding 18% tax to base_price
price_df = df.withColumn("final_price", col("base_price") * 1.18)

# Display the required columns
price_df.select("product_name", "base_price", "final_price").show(5)
```

### Q12 - Parquet Processing  
Read Parquet file, remove rows with null user_id, and save as CSV.

```python
# The Parquet file was already loaded at the beginning of the notebook

# Filter rows where user_id is not null
filtered_df = parquet_df.filter(col("user_id").isNotNull())

# Save the filtered DataFrame as a CSV file
filtered_df.write \
    .mode("overwrite") \
    .option("header", True) \
    .csv("output-csv")

print("Filtered data saved successfully.")

# Total number of rows in the filtered DataFrame after removing null values in the user_id column
print(filtered_df.count())
```

### Q14 - OR Filtering  
Filter rows where region is 'North' or priority is 'High'.

```python
# Filter rows where region is 'North' OR priority is 'High'
region_priority_df = df.filter((col("region") == "North") | (col("priority") == "High"))

# Display the filtered records
region_priority_df.show(5)
```
